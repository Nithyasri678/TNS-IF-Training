```python
class mob:
  def __init__ (self,brand,price):
    self.brand=brand
    self.price=price
    print(self.brand,"=",self.price)

m1=mob("iphone",90000)
m2=mob("Samsung",5000)
print(m1.brand,"=",m1.price)
print(m2.brand,"=",m2.price)
```
